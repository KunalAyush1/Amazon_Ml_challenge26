```python
from pathlib import Path
import sys

import pandas as pd


PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SRC_DIR = PROJECT_ROOT / "src"

if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

TRAIN_DIR = PROJECT_ROOT / "data" / "train"

print("Project root:", PROJECT_ROOT)
```

```python
from business_entity_resol.io.reader import read_source_file
from business_entity_resol.preprocessing.normalization import (
    normalize_dataframe,
)
```

```python
train_source1 = read_source_file(
    TRAIN_DIR / "train_source1.tsv"
)

print(train_source1.shape)
```

```python
sample = train_source1.sample(
    n=1000,
    random_state=42,
)

sample.head()
```

```python
normalized_sample = normalize_dataframe(sample)

print(normalized_sample.shape)
normalized_sample.head()
```

```python
display(
    normalized_sample[
        [
            "entity_id",
            "name_raw",
            "name_norm",
            "name_compact",
            "name_tokens",
        ]
    ].head(30)
)
```

```python
display(
    normalized_sample[
        [
            "entity_id",
            "address_raw",
            "address_norm",
            "address_compact",
            "address_tokens",
            "address_numbers",
        ]
    ].head(30)
)
```

```python
punctuation_examples = train_source1[
    train_source1["business_name"].str.contains(
        r"[.,&()/'-]",
        regex=True,
        na=False,
    )
].head(50)

punctuation_normalized = normalize_dataframe(
    punctuation_examples
)

display(
    punctuation_normalized[
        [
            "name_raw",
            "name_norm",
            "name_compact",
            "name_tokens",
        ]
    ]
)
```

```python
numeric_examples = train_source1[
    train_source1["business_address"].str.contains(
        r"\d",
        regex=True,
        na=False,
    )
].head(50)

numeric_normalized = normalize_dataframe(
    numeric_examples
)

display(
    numeric_normalized[
        [
            "address_raw",
            "address_norm",
            "address_numbers",
        ]
    ]
)
```

```python
train_source2 = read_source_file(
    TRAIN_DIR / "train_source2.tsv"
)

missing_address_examples = train_source2[
    train_source2["business_address"]
    .astype("string")
    .str.strip()
    .eq("")
].head(20)

missing_normalized = normalize_dataframe(
    missing_address_examples
)

display(
    missing_normalized[
        [
            "business_name",
            "address_raw",
            "address_norm",
            "address_tokens",
            "address_numbers",
        ]
    ]
)
```
